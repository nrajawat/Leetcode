```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next



class Solution:
    def mergeTwoLists(self, list1, list2):
        dummy = ListNode()
        cur = dummy

        while list1 and list2:
            if list1.val < list2.val:
                cur.next = list1
                list1 = list1.next
            else:
                cur.next = list2
                list2 = list2.next
            cur = cur.next


        if list1:
            cur.next = list1
        elif list2:
            cur.next = list2

        return dummy.next
```

```python
# list1: 1 -> 2 -> 4
list1 = ListNode(1)
list1.next = ListNode(2)
list1.next.next = ListNode(4)

# list2: 1 -> 3 -> 4
list2 = ListNode(1)
list2.next = ListNode(3)
list2.next.next = ListNode(4)

sol = Solution()
merged = sol.mergeTwoLists(list1, list2)

# Print merged list
while merged:
    print(merged.val, end=" ")
    merged = merged.next
```

```python
# Empty linked lists
list1 = None
list2 = None

sol = Solution()
merged = sol.mergeTwoLists(list1, list2)

print(merged)
```
