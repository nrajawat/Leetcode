```python
Given an integer x, return true if x is a palindrome, and false otherwise.

Example 1:
Input: x = 121
Output: true
Explanation: 121 reads as 121 from left to right and from right to left.

Example 2:
Input: x = -121
Output: false
Explanation: From left to right, it reads -121. From right to left, it becomes 121-. Therefore it is not a palindrome.

Example 3:
Input: x = 10
Output: false
Explanation: Reads 01 from right to left. Therefore it is not a palindrome.
 

Constraints:
-231 <= x <= 231 - 1
```

```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        n = str(x)
        rev = n[::-1]

        if n == rev:
            return True

        return False
        
```

```python

class Solution:
    def palindrome(self, x):
        t = x
        s = 0

        while(x > 0):
            r = x % 10
            s = s * 10 + r
            x = x // 10

        if s == t:
            return True
        else:
            return False
        
```

```python
sol = Solution()
x = 121
res = sol.palindrome(x)
print(res)
```

```python
sol = Solution()
x = -121
res = sol.palindrome(x)
print(res)
```

```python
sol = Solution()
x = 10
res = sol.palindrome(x)
print(res)
```
