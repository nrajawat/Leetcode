```python
Given a signed 32-bit integer x, return x with its digits reversed. If reversing x causes the value to go outside the signed 32-bit integer range [-231, 231 - 1], then return 0.
Assume the environment does not allow you to store 64-bit integers (signed or unsigned).


Example 1:
Input: x = 123
Output: 321

Example 2:
Input: x = -123
Output: -321

Example 3:
Input: x = 120
Output: 21
 

Constraints:
-231 <= x <= 231 - 1
```

```python

class Solution:
    def reverse(self, x):
        INT_MAX = 2**31 - 1
        INT_MIN = -2**31

        rev_x = 0
        sign = 1 if x >0 else -1
        x = abs(x)

        while x != 0:
            r = x%10
            x //= 10

            if rev_x > (INT_MAX - r) // 10:
                return 0

            rev_x = rev_x * 10 + r

        return sign * rev_x
        
```

```python
sol = Solution()
x = -123
res = sol.reverse(x)
print(res)
```
