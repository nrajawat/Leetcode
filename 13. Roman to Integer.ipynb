```python
class Solution:
    def romanToInt(self, s):

        roman = {
            'I' : 1,
            'V' : 5,
            'X' : 10,
            'L' : 50,
            'C' : 100,
            'D' : 500,
            'M' : 1000
        }

        res = 0
        prev_val = 0

        for char in reversed(s):
            cur_val = roman[char]

            if cur_val < prev_val:
                res -= cur_val
            else:
                res += cur_val

            prev_val = cur_val
            
        return res
```

```python
sol = Solution()
s = "III"
res = sol.romanToInt(s)
print(res)
```

```python
sol = Solution()
s = "LVIII"
res = sol.romanToInt(s)
print(res)
```
