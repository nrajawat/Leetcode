```python
class Solution:
    def longestValidParentheses(self, s):
        stack = [-1]
        lmax = 0

        for i, char in enumerate(s):
            if char == '(':
                stack.append(i)
            else:
                stack.pop()
                if not stack:
                    stack.append(i)
                else:
                    lmax = max(lmax, i-stack[-1])

        return lmax

        
```

```python
sol = Solution()
s = "(()"
res = sol.longestValidParentheses(s)
print(res)
```

```python
sol = Solution()
s = ")()())"
res = sol.longestValidParentheses(s)
print(res)
```

```python
sol = Solution()
s = ""
res = sol.longestValidParentheses(s)
print(res)
```
