```python
class Solution:
    def letterCombinations(self, d):
        if not d:
            return []

        phone = {
            '2': ['a', 'b', 'c'],
            '3': ['d', 'e', 'f'],
            '4': ['g', 'h', 'i'],
            '5': ['j', 'k', 'l'],
            '6': ['m', 'n', 'o'],
            '7': ['p', 'q', 'r', 's'],
            '8': ['t', 'u', 'v'],
            '9': ['w', 'x', 'y', 'z']
        }

        def backtrack(comb, nd):
            if not nd:
                output.append(comb)
            else:
                for letter in phone[nd[0]]:
                    backtrack(comb + letter, nd[1:])

        output = []
        backtrack('', d)
        return output
```

```python
sol = Solution()
digits = "23"
print(sol.letterCombinations(digits))
```
