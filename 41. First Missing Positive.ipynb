```python

class Solution:
    def firstMissingPositive(self, nums):
        n = len(nums)

        for i in range(n):
            while 1 <= nums[i] <= n and nums[i] != nums[nums[i]-1]:
                nums[nums[i] - 1], nums[i] = nums[i], nums[nums[i]-1]

        for i in range(n):
            if nums[i] != i+1:
                return i + 1

        return n + 1
```

```python

sol = Solution()
nums = [1, 2, 0]
res = sol.firstMissingPositive(nums)
print(res)
```
