```python
class Solution:
    def permutations(self, nums):
        if len(nums) == 0:
            return [[]]

        perms = self.permutations(nums[1:])
        res = []
        for p in perms:
            for i in range(len(p) + 1):
                pcopy = p.copy()
                pcopy.insert(i, nums[0])
                res.append(pcopy)
        return res
        
```

```python
sol = Solution()
nums = [1, 2, 3]
res = sol.permutations(nums)
print(res)
```

```python
sol = Solution()
nums = [0, 1]
res = sol.permutations(nums)
print(res)
```
