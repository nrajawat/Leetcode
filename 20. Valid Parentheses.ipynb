```python
class Solution:
    def isValid(self, s):
        st = []
        br = {')':'(', ']':'[', '}':'{'}

        for char in s:
            if char in br.values():
                st.append(char)
            elif char in br.keys():
                if not st or st[-1] != br[char]:
                    return False
                st.pop()

            else: 
                return False

        return not st
            
```

```python
sol = Solution()
s = "()"
print(sol.isValid(s))
```

```python
# Example 2
sol = Solution()
s = "()[]{}"
print(sol.isValid(s))
```
