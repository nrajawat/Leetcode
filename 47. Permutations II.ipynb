```python

class Solution:
    def permutations(self, nums):
        res = []
        perm = []
        count = { n:0 for n in nums}

        for num in nums:
            count[num] += 1

        def dfs():
            if len(perm) == len(nums):
                res.append(perm.copy())

                #brute force

            for num in count:
                if count[num] > 0:
                    perm.append(num)
                    count[num] -= 1
                    dfs()
                    count[num] += 1
                    perm.pop()
        dfs()
        return res
```

```python
sol = Solution()
nums = [1, 1, 2]
res = sol.permutations(nums)
print(res)
```
