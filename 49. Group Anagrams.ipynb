```python
class Solution:
    def groupAnagrams(self, strs):
        ana = {}

        for word in strs:
            sw = ''.join(sorted(word))
            if sw in ana:
                ana[sw].append(word)
            else:
                ana[sw] = [word]

        return list(ana.values())
```

```python
sol = Solution()
strs = ["eat","tea","tan","ate","nat","bat"]
res = sol.groupAnagrams(strs)
print(res)
```

```python
sol = Solution()
strs = [""]
res = sol.groupAnagrams(strs)
print(res)
```
