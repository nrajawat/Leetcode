```python

class Solution:
    def reverseKGroup(self, head, k):
        def reverseGroup(prev, cur, k):
            tail = prev
            while k > 0:
                next_node = cur.next
                cur.next = prev
                prev = cur
                cur = next_node
                k -= 1
            return prev, tail

        def getLength(node):
            l = 0
            while node:
                l += 1
                node = node.next
            return l

        l = getLength(head)
        if l < k:
            return head

        dummy = ListNode(0)
        dummy.next = head
        prev_group_end = dummy
        cur = head

        while l >= k:
            group_start = cur
            group_end = cur

            for i in range(k - 1):
                group_end = group_end.next

            next_group_start = group_end.next  # Save the start of the next group
            group_end.next = None  # Disconnect the current group

            # Reverse the current group
            reversed_start, reversed_end = reverseGroup(group_start, cur, k)

            prev_group_end.next = reversed_start
            reversed_end.next = next_group_start

            prev_group_end = group_start
            curr = next_group_start
            l -= k

        return dummy.next
        
                
```
