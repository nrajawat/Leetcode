```python
from collections import Counter

class Solution:
    def findSubstring(self, s: str, words: list) -> list:
        if not s or not words or not words[0]:
            return []

        word_len = len(words[0])
        num_words = len(words)
        total_len = word_len * num_words
        
        word_count = Counter(words)
        result = []
        
        # Traverse the string with a sliding window approach
        for i in range(word_len):
            left = i
            right = i
            current_count = 0
            current_window = Counter()

            while right + word_len <= len(s):
                word = s[right:right + word_len]
                right += word_len

                # If the word is in words, add it to the window counter
                if word in word_count:
                    current_window[word] += 1
                    current_count += 1
                    
                    # If we've added more occurrences of a word than allowed, move the left pointer
                    while current_window[word] > word_count[word]:
                        left_word = s[left:left + word_len]
                        current_window[left_word] -= 1
                        current_count -= 1
                        left += word_len
                    
                    # If we've matched all words, add the current left index to the result
                    if current_count == num_words:
                        result.append(left)
                else:
                    # Reset the window if the word is not part of the words list
                    left = right
                    current_window.clear()
                    current_count = 0
        
        return result
```
