```python

class Solution:
    def threeSumClosest(self, nums, target):
        nums.sort()
        csum = float('inf')

        for i in range(len(nums) - 2):
            l, r = i + 1, len(nums) - 1

            while l < r:
                total = nums[i] + nums[l] + nums[r]
                if abs(target - total) < abs(target - csum):
                    csum = total

                if total < target:
                    l += 1

                elif total >  target:
                    r -= 1

                else:
                    return csum

        return csum
```

```python
sol = Solution()
nums = [-1,2,1,-4]
target = 1
res = sol.threeSumClosest(nums, target)
print(res)
```
