```python
class Solution:
    def countAndSay(self, n):
        if n == 1:
            return "1"

        def generate_next(ter):
            res = []
            c = 1

            for i in range(1, len(term)):
                if term[i] == term[i-1]:
                    c += 1
                else:
                    res.append(str(c) + term[i-1])
                    c = 1

            res.append(str(c) + term[-1])
            return ''.join(res)

        term = "1"
        for _ in range(2, n+1):
            term = generate_next(term)
        return term
```

```python
sol = Solution()
n = 4
res = sol.countAndSay(n)
print(res)
```
