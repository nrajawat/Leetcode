```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def swapPairs(self, head):
        dummy = ListNode(0)
        dummy.next = head
        cur = dummy

        while cur.next and cur.next.next:
            fnode = cur.next
            snode = cur.next.next
            fnode.next = snode.next
            snode.next = fnode
            cur.next = snode
            cur = cur.next.next

        return dummy.next
        
```

```python
# Create linked list: 1 -> 2 -> 3 -> 4
head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)
head.next.next.next = ListNode(4)

sol = Solution()
result = sol.swapPairs(head)

# Print result
while result:
    print(result.val, end=" ")
    result = result.next
```
