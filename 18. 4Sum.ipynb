```python
class Solution:
    def fourSum(self, nums, target):
        nums.sort()
        res = []

        def Ksum(k, target, s, e, cur):
            if k == 2:
                while s < e:
                    total = nums[s] + nums[e]

                    if total == target:
                        res.append(cur + [nums[s], nums[e]])

                        while s < e and nums[s] == nums[s + 1]:
                            s += 1

                        while s < e and nums[e] == nums[e - 1]:
                            e -= 1

                        s += 1
                        e -= 1

                    elif total < target:
                        s += 1
                    else:
                        e -= 1

            else:
                for i in range(s, e - k + 2):
                    if i > s and nums[i] == nums[i - 1]:
                        continue

                    # Pruning
                    if nums[i] * k > target:
                        break

                    if nums[e] * k < target:
                        break

                    Ksum(
                        k - 1,
                        target - nums[i],
                        i + 1,
                        e,
                        cur + [nums[i]]
                    )

        n = len(nums)
        Ksum(4, target, 0, n - 1, [])

        return res
```

```python
sol = Solution()
nums = [1,0,-1,0,-2,2]
target = 0
res = sol.fourSum(nums, target)
print(res)
```
