```python

class Solution:
    def generateParenthesis(self, n):
        def generate_helper(cur_str, oc, cc):
            if oc == n and cc == n:
                res.append(cur_str)
                return

            if oc < n:
                generate_helper(cur_str + '(', oc+1, cc)

            if cc < oc:
                generate_helper(cur_str + ')', oc , cc+1)

        res = []
        generate_helper('', 0, 0)
        return res
                
        
```

```python
sol = Solution()
n = 3
res = sol.generateParenthesis(n)
print(res)
```
