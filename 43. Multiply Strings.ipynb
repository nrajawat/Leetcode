```python

class Solution:
    def multiply(self, num1, num2):
        res = ""

        res = int(num1) * int(num2)
        return str(res)
```

```python
sol = Solution()
num1 = "2"
num2 = "3"

res = sol.multiply(num1, num2)
print(res)
```
