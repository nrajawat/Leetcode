```python

class Solution:
    def maxArea(self, h):
        left = 0
        right = len(h) - 1
        max_area = 0

        while left < right:
            cur_area = min(h[left], h[right]) * (right - left)
            max_area = max(cur_area, max_area)

            if h[left] < h[right]:
                left += 1

            else:
                right -= 1

        return max_area
            
```

```python
sol = Solution()
height = [1,8,6,2,5,4,8,3,7]
res = sol.maxArea(height)
print(res)
```
