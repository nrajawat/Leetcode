```python

class Solution:
    def combinationSum2(self, cand, target):

        def bc(s, target, cur_comb):
            if target == 0:
                res.append(cur_comb[:])
                return 

            if target < 0 or s == len(cand):
                return

            for i in range(s, len(cand)):
                if i > s and cand[i] == cand[i-1]:
                    continue

                cur_comb.append(cand[i])
                bc(i+1, target-cand[i], cur_comb)
                cur_comb.pop()

        cand.sort()
        res = []
        bc(0, target, [])
        return res
```

```python

sol = Solution()
candidates = [10,1,2,7,6,1,5]
target = 8
res = sol.combinationSum2(candidates, target)
print(res)
```
