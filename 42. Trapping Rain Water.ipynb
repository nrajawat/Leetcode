```python

class Solution:
    def trap(self, h):
        if not h:
            return 0

        lmax = [0] * len(h)
        rmax = [0] * len(h)

        lmax[0] = h[0]
        for i in range(1, len(h)):
            lmax[i] = max(lmax[i-1], h[i])

        rmax[-1] = h[-1]
        for i in range(len(h)-2, -1, -1):
            rmax[i] = max(rmax[i+1], h[i])

        tw = 0
        for i in range(len(h)):
            tw += min(lmax[i], rmax[i]) - h[i]

        return tw
```

```python
sol = Solution()
height = [0,1,0,2,1,0,1,3,2,1,2,1]
res = sol.trap(height)
print(res)
```
