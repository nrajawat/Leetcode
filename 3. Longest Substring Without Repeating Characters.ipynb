```python

class Solution:
    def lengthOfLongestSubstring(self, s):
        n = len(s)
        res = 0
        char = set()
        l = 0

        for r in range(n):
            while s[r] in char:
                char.remove(s[l])
                l += 1

            char.add(s[r])
            res = max(res, r-l+1)

        return res
```

```python
s = "abcabcbb"

sol = Solution()
print(sol.lengthOfLongestSubstring(s))
```

```python
s = "bbbbb"

sol = Solution()
print(sol.lengthOfLongestSubstring(s))
```
