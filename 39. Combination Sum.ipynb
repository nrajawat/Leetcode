```python
class Solution:
    def combSum(self, cand, target):
        def bc(s, path, target):
            if target  == 0:
                op.append(path[:])
                return
    
            if target < 0:
                return
    
            for i in range(s, len(cand)):
                path.append(cand[i])
                bc(i, path, (target-cand[i]))
                path.pop()

        op = []
        bc(0, [],target)
        return op
    
            
```

```python
sol = Solution()
candidates = [2,3,6,7]
target = 7
res = sol.combSum(candidates, target)
print(res)
```
