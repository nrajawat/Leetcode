```python

class Solution:
    def divide(seld, dividend, divisor):
        if divisor == 0:
            return float('inf')

        if dividend == -2**31 and divisor == -1:
            return 2**31 - 1

        neg = (dividend < 0) != (divisor < 0) 
        dividend, divisor = abs(dividend), abs(divisor)
        q = 0

        while dividend >= divisor:
            temp, mul = divisor, 1

            while dividend >= (temp << 1):
                temp <<= 1
                mul <<= 1

            dividend -= temp
            q += mul

        if neg:
            q = -q

        return max(min(q, 2**31 - 1), -2**31)
            

            
```

```python
sol = Solution()
dividend = 10
divisor = 3
res = sol.divide(dividend, divisor)
print(res)
```
