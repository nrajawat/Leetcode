```python

class Solution:
    def longestCommonPrefix(self, s):
        if not s:
            return ""

        n = len(s)
        res = s[0]

        for i in range(1, n):
            j = 0
            while j < len(res) and j < len(s[i]) and res[j] == s[i][j]:
                j += 1
            res = res[:j]
            if not res:
                return ""

        return res
```

```python
sol = Solution()
s = ["flower","flow","flight"]
res = sol.longestCommonPrefix(s)
print(res)
```
