```python
class Solution:
    def searchInsert(self, nums, target):
        l, r = 0, len(nums) - 1

        while l <= r:
            mid = l + (r - l) // 2

            if nums[mid] == target:
                return mid

            elif nums[mid] < target:
                l = mid + 1

            else:
                r = mid - 1

        return l
```

```python
sol = Solution()
nums = [1,3,5,6]
target = 5
res = sol.searchInsert(nums, target)
print(res)
```
