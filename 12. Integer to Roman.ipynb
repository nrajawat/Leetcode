```python

class Solution:
    def intToRoman(self, num):
        roman = {
            1000: 'M', 900: 'CM', 500:'D', 400:'CD',
            100:'C', 90:'XC', 50:'L', 40:'XL',
            10:'X', 9: 'IX', 5:'V', 4:'IV', 1: 'I'
        }

        res = ' '
        for val, sym in roman.items():
            while num >= val:
                res += sym
                num -= val

        return res
```

```python
sol = Solution()
num = 3749
res = sol.intToRoman(num)
print(res)
```
