```python
class Solution:
    def myAtoi(self, s):
        INT_MAX = 2**31 - 1
        INT_MIN = -2**31

        s = s.strip()
        if not s:
            return 0

        sign = 1
        if s[0] in ('+', '-'):
            if s[0] == '-':
                sign = -1
            s = s[1:]

        res = 0
        for char in s:
            if char.isdigit():
                d = int(char)
                if res > (INT_MAX - d) //10:
                    return INT_MAX if sign == 1 else INT_MIN
                res = res * 10 + d
            else:
                break
                
        return sign * res
```

```python
sol = Solution()
s = "42"
res = sol.myAtoi(s)
print(res)
```
