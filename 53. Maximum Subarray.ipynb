```python

class Solution:
    def maxSubArray(self, nums):
        if not nums:
            return 0

        max_sum = nums[0]
        cur_sum = nums[0]

        for i in range(1, len(nums)):
            num = nums[i]
            cur_sum = max(num, cur_sum + num)
            max_sum = max(max_sum, cur_sum)

        return max_sum
```

```python
sol = Solution()
nums = [-2,1,-3,4,-1,2,1,-5,4]
res = sol.maxSubArray(nums)
print(res)
```
