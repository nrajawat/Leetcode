```python
class Solution:
    def jump(self, nums):
        if len(nums) == 1:
            return 0

        max_reach = nums[0]
        next_max_reach = nums[0]
        jumps = 1

        for i in range(1, len(nums)):
            if i > max_reach:
                jumps += 1
                max_reach = next_max_reach
            next_max_reach = max(next_max_reach, i+nums[i])
        return jumps
```

```python
sol = Solution()
nums = [2, 3, 1, 1, 4]
res = sol.jump(nums)
print(res)
```
