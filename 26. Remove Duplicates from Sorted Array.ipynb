```python
class Solution:
    def removeDuplicates(self, nums):
        if not nums:
            return 0

        uniq_idx = 0

        for i in range(1, len(nums)):
            if nums[i] != nums[uniq_idx]:
                uniq_idx += 1
                nums[uniq_idx] = nums[i]

        return uniq_idx + 1
```

```python
sol = Solution()
nums = [1,1,2]
res = sol.removeDuplicates(nums)
print(res)
```
