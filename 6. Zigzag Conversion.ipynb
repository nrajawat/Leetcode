```python
class Solution:
    def convert(self, s, numrows):
        if numrows == 1: return s

        res = ""
        for r in range(numrows):
            inc = 2*(numrows - 1)

            for i in range(r, len(s), inc):
                res += s[i]
                if (r > 0 and r < numrows-1 and i+inc - 2*r < len(s)):
                    res += s[i+ inc - 2*r]

        return res
```

```python
sol = Solution()
res = sol.convert("PAYPALISHIRING", 3)
print(res)
```

```python
sol = Solution()
res = sol.convert("PAYPALISHIRING", 4)
print(res)
```
