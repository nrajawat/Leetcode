```python
from typing import List

class Solution:
    def searchRange(self, nums, target):
        def bs(ff):
            l, r = 0, len(nums)-1
            res = -1

            while l <= r:
                mid = l + (r - l) // 2
                if nums[mid] == target:
                    res = mid
                    if ff:
                        r = mid - 1

                    else:
                        l = mid + 1

                elif nums[mid] < target:
                    l = mid + 1

                else:
                    r = mid - 1
                    
            return res

        fo = bs(ff = True)
        if fo == -1:
            return [-1, -1]

        lo = bs(ff = False)
        return [fo, lo]
```

```python
sol = Solution()
nums = [5,7,7,8,8,10]
target = 8
res = sol.searchRange(nums, target)
print(res)
```
