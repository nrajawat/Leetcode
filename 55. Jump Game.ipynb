```python

class Solution:
    def canJump(self, nums):
        max_reach = 0
        n = len(nums)

        for i in range(n):
            if i > max_reach:
                return False

            max_reach = max(max_reach, i + nums[i])
            if max_reach >= n - 1:
                return True

        return False
```

```python
sol = Solution()
nums = [2,3,1,1,4]
print(sol.canJump(nums))
```

```python
sol = Solution()
nums = [3,2,1,0,4]
print(sol.canJump(nums))
```
