```python
class Solution:
    def twoSum(self, num, target):
        n = len(nums)
        res = {}

        for i in range(n):
            comp = target - nums[i]
            if comp in res:
                return [res[comp], i]

            res[nums[i]] = i

        return []

```

```python
nums = [2, 7, 11, 15]
target = 9

obj = Solution()
print(obj.twoSum(nums, target))
```
